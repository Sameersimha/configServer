```python
  from pandas.io.formats import style
  import dash
  from dash import dcc, html, Input, Output
  import pandas as pd
  import plotly.express as px

  #Load the sample excel file
  df = pd.read_excel("/content/Project-Management-Sample-Data.xlsx", sheet_name='Project Management Data')

  # initialize dash app
  app = dash.Dash(__name__)

  #Define The App Layout
  app.layout = html.Div([
      html.H1("Project Management Dashboard", style={'textAlign': 'center', 'font-family':'Times New Roman'}),

      html.Div([
          html.Div([
              html.Label("Select Project:"),
              dcc.Dropdown(
                  id='Project-dropdown',
                  options=[{'label': project, 'value': project} for project in df['Project Name'].unique()],
                  value=df['Project Name'].unique()[0],
                  clearable=False
              )
            ], style={'width': '30%', 'display': 'inline-block', 'padding':'10px'}),

          html.Div([
              html.Label("Select Manager:"),
              dcc.Dropdown(
                  id='Manager-dropdown',
                  options=[{'label': manager, 'value': manager} for manager in df['Assigned to'].unique()],
                  value=df['Assigned to'].unique()[0],
                  clearable=False
              )
          ], style={'width': '30%', 'display': 'inline-block', 'float': 'right' ,'padding':'10px'})
      ], style={'display': 'flex', 'justify-content': 'space-around'}),
      html.Div([
          html.Div([
              dcc.Graph(id='project-graph')
          ], style={'display': 'flex', 'justify-content':'space-around'})
      ])
  ])

  @app.callback(
      Output('project-graph', 'figure'),
      [Input('Project-dropdown', 'value')]
  )
  def update_graph(selected_project):
      filtered_df = df[(df['Project Name'] == selected_project)]
      fig1 = px.bar(filtered_df, x='Assigned to', y='Days Required', title=f"Tasks Assigned", barmode='group')
      return fig1


  if __name__ == '__main__':
      app.run(debug=True)
```
